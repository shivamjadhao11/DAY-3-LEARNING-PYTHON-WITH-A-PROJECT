```python
import csv
with open("emp.csv","w",newline='') as f:
  w=csv.writer(f)
  w.writerow(["enO","ename","esalary","eaddr"])
  n=int(input("enter number of employes:"))
  for i in range(n):
        eno=input("enter employees no:")
        ename=input("enter employee name:")
        esalary=input("enter employee salary:")
        eaddr=input("enter employee address:")
        w.writerow([eno,ename,esalary,eaddr])
  print("Data inseted successfully")
```

```python
import numpy as np
students=np.array([
    [85,90,78],
    [30,21,44],
    [95,92,88],
    [40,50,45],
    [72,68,75],
])
students
```

```python
avg_marks=np.mean(students,axis=1)
print("Average marks:",avg_marks)
```

```python
avg_marks=np.mean(students,axis=0)
print("Average marks:",avg_marks)
```

```python
topper=np.argmax(avg_marks)
print("Topper student:",topper)
```

```python
failed_student=np.where(avg_marks<50)
print("Failed student:",failed_student[0]+1)
```

```python
for i, avg in enumerate(avg_marks):
    status="pass" if avg>50 else "Fail"
    print(f"student {i+1} Avg = {avg:.2f} | {status}")
```

```python
months=['jan','feb','mar','april']
sales=[250,300,350,500]
plt.bar(months,sales,color='red')
plt.title('monthly sales')
plt.xlabel('months')
plt.ylabel('sales')
plt.show()
```

```python
plt.barh(months,sales,color='green')
plt.title('monthly sales(Horozontal)')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```
